# What is an array?

*   An array is simply a collection of multiple values stored together.
*   Data structure in programming
*   NumPy array: stores numerical values together and lets you perform          mathematical operations on them efficiently.



```python
np.arange(6)
```

```python
a = np.arange(6)
print(a)
a.shape # only 6 col
```

#2D array

```python
a2 = a[:, np.newaxis]
print(a2)
a2.shape # 6 rows 1 col 2d arr
```

```python
a2 = a[np.newaxis, :]
print(a2)
a2.shape # 6 col 1 row
```

#3D array

```python
a3 = a2[np.newaxis, :]
print(a3)
a3.shape # 1 block 1 row 6 val; block-> 3 levels of brackets
# axis 0 has size 1, axis 1 has size 1, axix 2 has size 6
```

# why do we have 1d, 2d, 3d arrays and where we need in data science and ml?

We use different dimensions because different types of data have different structures. For a simple list, we use 1D. For rows and columns like a table, we use 2D. For data with an additional level, we use 3D or more. This helps us organize data properly and allows ML models and other programs to process the data correctly

* **1D array** → A simple list of values. Example: [10, 20, 30, 40]. Used for one type of data, like marks, ages, or a list of numbers.
* **2D array **→ Data arranged in rows and columns, like a table. Example: [[10, 20], [30, 40]]. Very common in Data Science and ML for datasets.
* **3D array** → Data arranged in multiple tables/blocks. Commonly used for images (height × width × color channels) and some ML data.
* **Why do we need different dimensions?** → Because real-world data has different structures. A simple list needs 1D, a table needs 2D, and more complex data like images or batches may need 3D+.
* **In ML:** NumPy arrays help us store, organize, and perform calculations on numerical data before giving that data to ML models.
